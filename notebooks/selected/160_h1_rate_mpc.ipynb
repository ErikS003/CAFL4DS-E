# N160 · One-step learning-rate control with median gain floors

Each arm gets 15,232 decisions in each of two BDD streams, 30,464 total. Start
from the same warm MAE and fresh AdamW. Use the N159 streams and its already
decoded frames. The current-image reconstruction loss and contextual RankMe
define the objective. The recent assay uses only observed frames and the current
frame. All positive actions train on the current image alone.

At each arrival forecast the immediate outcome of skip, 1/16 LR, and 1/4 LR.
Choose an eligible positive rate by minimizing predicted/current loss plus half
of current/predicted RankMe. Skip has score 1.5 and preserves the learner.
Every update must predict strictly positive improvement on both metrics. A
calibrated metric must exceed the median measured signed gain for that rate
on N158's 106 offline calibration origins, floored at zero. This gives the
four requested guards: RankMe only, loss only, both, and neither. Policies
with identical thresholds share one deterministic trajectory per source pool.
The reference arms accept every post-fill image at nominal or quarter rate.

Use only N158's offline immediate-action heads. Update each head online after
its rate is executed and its outcome measured. The simple next-arrival H1
regressor remains a separate predictor trained online from the last reading
of each metric. Report both predictors against persistence on matched targets.
Actual incoming loss, contextual RankMe, accepted-image loss and fixed-audit
health are recorded over time. Per-clip plots summarize the exact image-level
ledgers. The development archive and the offline calibration split are reused.
