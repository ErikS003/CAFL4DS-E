# N157 · Does replay timing help with a 128-arrival memory window?

The measured N156 oracle had limited room over historical replay at 16 updates.
N157 changes the decision timescale and limits access to the last 128 arrivals.
Within this experiment, compare timing against a steady plan with the same replay
dose. Across N156 and N157, the horizon, age limit and development pools differ.

## Frozen design

- The current frame receives half the training loss. The four replay frames
  split the other half. Eligible ages are 1–128 arrivals. Recent means ages 1–32.
  Older means ages 33–128. Each arrival enters memory after its own update.
- Two 96-update blocks use one of three allocations. History is four older.
  Mixed is two recent and two older. Steady 3:1 is one recent and three older.
  The five plans are steady history, steady mixed, steady 3:1, mixed then history,
  and history then mixed. The last two use exactly the same replay dose as steady
  3:1. Each plan gets 192 steps, 960 gradient images and LR 0.000025 per step.
  All plans then receive 96 common history updates with identical draws and seeds.
- Reuse the warm MAE with fresh AdamW. Use two 20-clip BDD training-video pools
  disjoint from N155 and N156. These videos were available from earlier local
  studies. Both pools balance highway and city clips within each lighting group.
  Each pool uses a 4-day/6-night/4-day/6-night order and one whole-clip shuffle
  with a common two-clip fill. The shuffle is chosen from a fixed seed sequence
  solely for feasibility of the frozen origin strata. The learner starts after
  the 192-frame fill. Frame order inside each clip is chronological.
- Use two paired training seeds (15701 and 15707). Each of eight reference paths
  follows fixed history replay and supplies four nonoverlapping 288-arrival
  windows. Two origins follow an observed lighting boundary by eight frames.
  Two have at least 128 same-light past frames. Lighting metadata stratifies
  results. Planner features use only the current frame and causal memory.
- A fixed 64-image seen panel measures retention and RankMe at H96, H192 and
  H288. It can overlap training replay. H288 also scores a separate panel of 16
  video clips balanced by day/night and highway/city. One frozen origin per
  reference path gets an independent scene probe after H288. Probes never choose
  plans. H16 and H96 incoming-loss readings are diagnostics. The primary outcome
  is mean incoming pre-update reconstruction through H192.

## Qualification

Choose the strongest steady comparator by incoming loss in the other clip pool.
Eligible plans improve H192 incoming loss by at least 0.0005, with fixed-bank
loss and RankMe within 0.0005 and 0.01 of that comparator. Among plans within
0.0001 of the lowest eligible loss, prefer higher RankMe. The H288 independent
audit uses the N156 tolerances of 0.001 loss and 0.02 RankMe, by pool and lighting.

First require an equal-pool mean guarded oracle gain of at least 0.0005, positive
gain in each pool, and H288 audit protection. Then require the five-plan oracle
to improve on an oracle limited to steady allocations by at least 0.00025 on
average and positively in both pools. Also report early/late pulses against
steady 3:1 directly. A direct ridge predictor is fitted only if both screens
pass. Its 90th-percentile margins use training origins excluded by target clip,
and it is evaluated in the other source pool. This is development qualification
with two source pools. The scene probe is an independent descriptive check.

Scientific budget is 32 origins, 160 plans, 39,936 physical branch updates and
13,824 reference updates. This includes the 96-update continuation for every
plan and reuse of the three common first-block states. Preflight is separate.
Each origin and the reference state are saved atomically.
