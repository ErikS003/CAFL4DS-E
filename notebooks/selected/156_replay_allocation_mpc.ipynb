# N156 · Can a replay allocation plan beat fixed memory replay?

N155's full historical reservoir improved held-out reconstruction over FIFO in all
12 matched comparisons. The blocked-minus-shuffled interaction was small and the
scene probe did not improve consistently. N154's calibrated selector stayed with
fixed replay. Here the decision is how much to rehearse from recent versus older
memory while giving each alternative the same learning rate and gradient work.

## Frozen collection

- Reuse both clip-disjoint N155 development pools and all three orders per pool.
  Use N155 seed 15501, the same warm RGB64 MAE, fresh AdamW, and the same 288-frame
  fill. Each reference trajectory runs fixed 2+2 replay through all 1,920 arrivals.
  The completed N155 256-slot historical trajectory is a separate benchmark.
- Keep 128 recent frames and an Algorithm R reservoir of 128 frames that aged
  out of recent memory. Admit each frame after its own update. All branches use
  the same admission, pair replay draws by slot and training views by arrival.
- Each of six trajectories supplies 12 frozen origins. Use three eligible
  illumination boundaries per order, with a plan beginning eight frames before
  each boundary and a separate plan 40 frames after it. Six stable origins have
  their whole assessment within one clip. The 48-frame windows are disjoint
  within each trajectory. Metadata stratifies offline collection. The planner
  receives the current frame, model and causal bank summaries only.
- Nine plans combine two eight-update blocks. A block draws four recent, two
  recent and two historical, or four historical replay images at each update.
  The current frame supplies half the loss. Every plan uses 16 optimizer steps,
  80 gradient images and LR 0.000025 per step. All branches then get the same
  32-update 2+2 continuation policy with identical replay IDs and seeds. Every
  fork starts from the same model and AdamW state at its origin. Frame identities
  become usable as they arrive. The reference trajectory is trained separately.

## Outcomes and qualification

The planning objective is mean incoming pre-update reconstruction through H16.
Terminal retention loss and RankMe use a fixed 64-image panel of already-seen
frames, 32 from each memory tier. This panel can overlap replay and is scored
again at H48. The independent eight-clip N155 audit is scored at H48 by lighting.
A prespecified 12-origin subset also gets separate held-out scene probes at H48
for the measured guarded oracle and its fixed comparator. Probe labels never
select plans. Origins, orders and assays share clips within each source pool.

Leave one whole pool out. Select the strongest constant allocation by incoming
loss in the other pool. Require a plan to improve H16 incoming loss by at least
0.0005 and stay within 0.0005 bank loss and 0.01 RankMe of that comparator.
Within 0.0001 of the best eligible loss, prefer higher RankMe. Use the fixed
comparator when no plan qualifies. Require a mean oracle gain of at least 0.0005
across pools, positive gain in each pool, and H48 audit protection in each pool
and lighting group (loss tolerance 0.001, RankMe tolerance 0.02).

Only if that screen passes, fit a direct ridge advantage predictor from origin
features in the other pool. Calibrate using one-sided 90th-percentile errors
from training origins whose target clips were excluded from their fitting set.
Test the calibrated selector against fixed replay on the held-out pool. The
selector must meet the same incoming gain and audit requirements with zero
observed H16 guard violations. This is exploratory qualification on two reused
development pools. Any live controller or online learning would be a later run.

Scientific budget is 72 origins, 648 plan outcomes, 27,648 branch updates
including continuations, and 9,792 reference updates. Engineering preflight is
separate. Each origin and reference checkpoints are saved atomically.
