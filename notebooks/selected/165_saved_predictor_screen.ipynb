Use the same paired states and outcomes for mean-per-rate, 14-feature, and gradient-preview predictors.
Keep N164 stopped. The screen uses saved outcomes. Ridge penalty 20, training-only standardization and clipping, and the N164 256-label forgetting half-life are inherited.

Prediction error concerns the difference from quarter rate in incoming pre-update loss averaged over the next 16 arrivals. Quarter effects are structurally zero. RMSE excludes that trivial action. Decisions choose the smallest predicted loss effect among skip, quarter and nominal.

Both feature groups are available on seven completed preview-controller trajectories, 833 paired origins. Base-only paths and the interrupted random control are excluded from the common-case cohort. The source states were visited by preview policies, so this is a conditional saved-state screen.

Offline splits contain 261 fit, 117 calibration and 117 test origins. Calibration cases are excluded from fitting in this screen. Offline and deployment videos are disjoint. Pool folds exclude every clip in the target pool from fitting. Adaptive forecasts use only outcomes that matured before the next origin.

Primary summaries weight clips equally within a pool and pools equally. Repeated seeds and controller paths share clips. The descriptive intervals resample clips jointly across repeats and do not establish capture-session independence or remove serial dependence.

Random controls preserve chosen-rate counts within each trajectory and target-video group. The matching oracle preserves the same counts. These are retrospective diagnostics on isolated one-action forks, with a common quarter continuation; gains do not establish a cumulative closed-loop improvement.

All data and several prediction results were previously inspected during N164. This is development screening.

## Findings and interpretation

14 features has held-out-pool gain -0.000065 versus the mean-effect policy and -0.000011 versus matched random. The predeclared directional screen does not pass.

14 + preview has held-out-pool gain -0.000002 versus the mean-effect policy and +0.000303 versus matched random. The predeclared directional screen does not pass.

Action-selection gains and regret determine whether context improves decisions. A lower forecast RMSE can coexist with worse rate choices.

Adaptive replay tests predictor updates on the recorded states and paired feedback. A controller that makes different choices would create a different subsequent learner trajectory, which this screen does not simulate.

The health outputs describe the same 16-frame panels used in N164. Their measurement limitations carry into this screen.

## Reading the result

The mean-effect policy selected nominal at all 833 sampled decision opportunities. The decision under study is followed by quarter-rate updates on intervening arrivals. This is a periodic-rate comparator on saved states.

Gradient preview reduced held-out-pool loss-effect RMSE by 25.4% relative to the mean-effect predictor, from 0.003016 to 0.002251.

Its mean loss gain versus that policy was -0.000002. The descriptive clip-resampling interval was [-0.000056, +0.000057]. Causal adaptation gave gain +0.000007, with opposite signs in the two pools.

At its own rate budget, preview achieved gain +0.000303 over matching randomized choices, positive in both pools. It selected 47 skips, 154 quarter rates and 632 nominal rates, compared with 833 nominal choices for the mean policy. Similar loss to a more aggressive constant selector and better timing at a lower rate dose are distinct findings. LR dose is not a measurement of energy or training FLOPs.

The 14-feature model had lower forecast RMSE than the mean-effect predictor, but its decisions increased loss by 0.000065 on average relative to the mean policy, with worse loss in each held-out pool.

The screen supports preview as a predictor and shows conditional timing information at its selected rate budget. It does not establish an incremental loss reduction over the simple periodic-rate comparator. A future closed-loop test should include that comparator at explicitly matched budgets.

These results use paired 16-arrival outcomes from seven completed preview-policy trajectories. The states and clips were previously inspected development material. Predictions and decisions are scored on the same saved states for all models. Different deployed decisions would generate different subsequent states.
